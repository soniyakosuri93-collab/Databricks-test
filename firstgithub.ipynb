```python
# Parse the raw data into typed rows (skip the header line)
from datetime import date
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DateType, DoubleType

rows = []
for line in employee_data.strip().split("\n")[1:]:
    emp_id, name, dept, join_date, salary, rating = line.split("\t")
    rows.append((
        int(emp_id),
        name,
        dept,
        date.fromisoformat(join_date),
        int(salary.replace(",", "")),
        float(rating),
    ))

schema = StructType([
    StructField("Employee_ID", IntegerType(), False),
    StructField("Name", StringType(), True),
    StructField("Department", StringType(), True),
    StructField("Join_Date", DateType(), True),
    StructField("Salary", IntegerType(), True),
    StructField("Rating", DoubleType(), True),
])

employees_df = spark.createDataFrame(rows, schema)
employees_df.createOrReplaceTempView("employees")
display(employees_df)
```

```python
from pyspark.sql import functions as F

result_df = (
    employees_df.select(
        F.col("Employee_ID").alias("employee_id"),
        F.col("Name").alias("name"),
        F.col("Department").alias("department_name"),
        F.col("Join_Date").alias("join_date"),
        F.col("Salary").alias("salary"),
        F.col("Rating").alias("rating"),
    )
    .orderBy(F.desc("salary"))
)
```
